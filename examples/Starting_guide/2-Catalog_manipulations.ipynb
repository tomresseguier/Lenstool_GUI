### Initialize PyQt and create a `fits_image` instance (see *Starting guide 1*).

```python
%gui qt5
from PyQt5.QtWidgets import QApplication
import sys
import os

app = QApplication.instance()
if app is None:
    app = QApplication(sys.argv)

from visualens.fits_image import fits_image

DATA_dir = os.path.join( os.path.dirname(os.getcwd()), 'DATA' )

im = fits_image(DATA_dir + "/RGB_cropped.fits")
im.boost()
```

---
## Import a catalog
> The `fits_image.import_catalog()` function allows the user to import astronomical catalogs. The input can be the path to a fits or raw text (csv) catalog, or a `astropoy.table.Table` object directly.

```python
im.import_catalog(DATA_dir + '/phot-eazy.cat')
```

## Catalog functions
- To access the catalog itself:
> `fits_image.imported_cat.cat`

```python
im.imported_cat.cat
```

- Plots the objects found in the catalog as selectable ellipses:
> `fits_image.imported_cat.plot()`

```python
im.imported_cat.plot(scale=8., color=[0,1,1], text_column=None, linewidth=3, marker=None)
```

- You can select/deselect individual objects by clicking on them, which prints their index. The resulting mask is accessible with:
> `im.imported_cat.selection_mask`

```python
i = 409
print(im.imported_cat.cat[i]['ra'])
print(im.imported_cat.selection_mask)
```

- Plots one column of the catalog as a label next to each object:
> `fits_image.imported_cat.plot_column()`

```python
im.imported_cat.plot_column('z_phot')
```

- Clears the plot:
> `fits_image.imported_cat.clear()`

```python
im.imported_cat.clear()
```

### Selection panel

- Creates a panel with a scatter plot of objects according to two specified columns:
> `fits_image.imported_cat.make_selection_panel()`

> **Press `Shift` + click and drag to create a selection box.** Selected galaxies are colored in red in both scatter plot and on the image. **Clicking on the galaxy on the image shows where it lies in the scatter plot.**

> The selection is saved in `im.imported_cat.selection_mask`. This is particularly useful when making a red sequence selection of cluster galaxies for example. You can save the selection directly with `fits_image.imported_cat.export_to_potfile(path)`.

- Creates a panel with a scatter plot of objects according to two specified columns:
> Close the panel with `fits_image.imported_cat.remove_selection_panel()`

```python
im.imported_cat.make_selection_panel(xy_axes=['z_phot', 'f200w_flux'])
```

```python
im.imported_cat.remove_selection_panel()
```
