### Initialize PyQt
> **Required**: Run these cells to create a PyQt app. PyQt is the backbone of visualens and allows much more optimized, faster and complex graphical abilities when compared to libraries like matplotlib.

```python
%gui qt5
from PyQt5.QtWidgets import QApplication
import sys
import os

app = QApplication.instance()
if app is None:
    app = QApplication(sys.argv)
```

### Import the `fits_image` class from `visualens`.
> This class is the core of visualens.

```python
from visualens.fits_image import fits_image
```

### Initialize work directories

```python
DATA_dir = os.path.join( os.path.dirname(os.getcwd()), 'DATA' )
```

### Create an instance of the `fits_image` class and initialize it with a FITS image.
> As a visualization tool, the image you import here will act as the canvas of most of visualens' functions. This image only serves visualization purposes and therefore the actual pixel values do not matter, which is why we suggest importing a pretty **RGB image**. As we will see in these notebooks, visualens does have photometric tools that use images from individual filters, in flux or counts units. We will see how to load such filters.

> All of visualens' functions are accessible through the instance of the fits_image class that you create initially. `.import()` and `.load()` functions allow the user to load their data in memory for visualization, interaction and processing.

> **The image will open in an independent PyQt window. Look in your Dock or Taskbar for a "python" window.** Later on, other windows will open in the same way. Only a few matplotlib plots created by visualens will appear in the notebook directly when relevant (most things are plotted directly into PyQt).

> We suggest placing the PyQt window next to your notebook so that you can see the effects of the different functions directly as you execute the rest of this notebook.

```python
im = fits_image(DATA_dir + "/RGB_cropped.fits")
im.boost()  # Optional: this increases the contrast of the image.
```

- Create a secondary window / recreate the primary window
> `fits_image.plot_image()`

```python
im.plot_image()
```
