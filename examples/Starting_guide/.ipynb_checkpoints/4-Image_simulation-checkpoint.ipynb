# Starting guide 4: **Image simulation with `lenstronomy`**

### Initialize PyQt, create a `fits_image` instance (see *Starting guide 1*) and import a lens model (see *Starting guide 3*).

```python
%gui qt5
import matplotlib
matplotlib.use('QtAgg')
%matplotlib qt
from PyQt5.QtWidgets import QApplication
import sys
import os

app = QApplication.instance()
if app is None:
    app = QApplication(sys.argv)

from visualens.fits_image import fits_image

DATA_dir = os.path.join( os.path.dirname(os.getcwd()), 'DATA' )

im = fits_image(DATA_dir + "/RGB_cropped.fits")
im.boost()

model_path = DATA_dir + "/lens_model/arc_optimization/"
im.import_lenstool(model_path, use_wrapper=True)
im.lt.set_lt_z(6.2)
im.lt.plot()
```

## Load images from individual filters

Image simulation and forward modeling require using physical flux unit pixels.

- To load fits images from different filters:
> `fits_image.load_filters()`

By default, visualens will look for fits images in a "FILTERS" directory placed in the same directory as the RGB image but you can specify the path with `filter_dir=path_to_dir`. For now, `visualens` only recognizes HST and JWST filters. The fits images must have the filter name either in their header or their filename (upper or lowercase unimportant).

```python
im.load_filters()
```

## Select the area to simulate and start building a source light model

In your primary image, **press `Shift` + click and drag to select the image plane area to simulate.** Then:

- To create the image simulation window:
> `fits_image.lt.start_simulate_image(filter=filter_name)`

```python
im.lt.start_simulate_image(which_filter='F200W')
```
