## Import a lens model

`visualens` uses `Lenstool`'s python wrapper as its lensing engine and takes as inputs lens model files in the `Lenstool` format.

- To import your lens model (`model_path` can be the path to the directory containing your lens model's files, or the path to any of these files):
> `fits_image.import_lenstool(model_path, use_wrapper=True)`

**Careful:** This will move the current working directory to your lens model's directory, so make sure to use absolute paths.

All lensing functions are now accessible in the `fits_image.lt` instance. **`visualens` automatically calculates the sources and predicted image positions from the multiple images, as well as lensing quantities like magnification, shear, etc.**, so this step can take a little bit to run if you have many multiple images (a few seconds, typically less than a minute).


> **Note:** `Lenstool`'s python wrapper will overwrite some of your output files (`image.all`, `ce.dat`, etc.) as it recomputes the corresponding data. If you do not want to use `Lenstool`'s python wrapper, you can use `use_wrapper=False`. Only the functionalities that do not depend on `Lenstool` will be available.

> If you do not provide the `use_wrapper` parameter, `visualens` will ask you what you want to do and offer to make a copy of your model's directory. If you choose the latter option, it will create and run in a directory with the same name as your model's directory, but with the `_SAFE` keyword appended to it.

```python
model_path = DATA_dir + "/lens_model/"
im.import_lenstool(model_path, use_wrapper=True)
```

## Set the source plane's redshift

- Many functions of `fits_image.lt` run calculations in the source plane. To set the source plane's redshift:
> `fits_image.lt.set_lt_z(redshift)`

> `visualens` will automatically compute the critical and caustic curves at the given redshift, which can take a little bit of time. One it has computed them, it saves them in `lt_curves.pkl` and `lt_caustics.pkl` to avoid recalculating them later.

```python
im.lt.set_lt_z(6.2)
```

## Load images from individual filters

Image simulation and forward modeling require using physical flux unit pixels.

- To load fits images from different filters:
> `fits_image.load_filters()`

By default, visualens will look for fits images in a "FILTERS" directory placed in the same directory as the RGB image but you can specify the path with `filter_dir=path_to_dir`. For now, `visualens` only recognizes HST and JWST filters. The fits images must have the filter name either in their header or their filename (upper or lowercase unimportant).

```python
im.load_filters()
```

```python
im.filters['F200W'].psf.plot()
```
